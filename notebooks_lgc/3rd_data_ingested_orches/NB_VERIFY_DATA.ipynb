```python
# ============================================================
# CLEANUP HISTORICAL POI WITH INCOMPLETE REQUIRED LANGUAGES
#
# Business rule:
#   Each POI group MUST contain all required languages:
#       vi + en + ko
#
# If one or more required languages are missing:
#   -> Deactivate ALL POIs in that poi_group_id
#   -> poi_master.is_active = false
#   -> ai_enriched_poi.is_active = false
#
# IMPORTANT:
#   - Detect based on active POIs in poi_master.
#   - Freeze/materialize invalid poi_id list BEFORE updating targets.
#   - Do NOT delete any data.
# ============================================================

from pyspark.sql import functions as F

# ============================================================
# CONFIG
# ============================================================

POI_MASTER = "sgr_visitvn_dev.lh_vv_gold.dbo.poi_master"
AI_ENRICHED_POI = "sgr_visitvn_dev.lh_vv_gold.dbo.ai_enriched_poi"

REQUIRED_LANGUAGES = ["vi", "en", "ko"]


# ============================================================
# 1. VALIDATE TARGET SCHEMA
# ============================================================

poi_master_cols = set(spark.table(POI_MASTER).columns)
ai_enriched_cols = set(spark.table(AI_ENRICHED_POI).columns)

required_master_cols = {
    "poi_id",
    "poi_group_id",
    "lang",
    "is_active"
}

missing_master_cols = required_master_cols - poi_master_cols

if missing_master_cols:
    raise Exception(
        f"{POI_MASTER} missing required columns: "
        f"{sorted(missing_master_cols)}"
    )

required_ai_cols = {
    "poi_id",
    "is_active"
}

missing_ai_cols = required_ai_cols - ai_enriched_cols

if missing_ai_cols:
    raise Exception(
        f"{AI_ENRICHED_POI} missing required columns: "
        f"{sorted(missing_ai_cols)}"
    )


# ============================================================
# 2. DETECT INVALID POI GROUPS
#
# Invalid if active group is missing ANY of:
#   vi
#   en
#   ko
#
# Example:
#   vi,en,ko    -> VALID
#   vi,en       -> INVALID
#   vi,ko       -> INVALID
#   en,ko       -> INVALID
#   vi,en,ko,ja -> VALID
# ============================================================

invalid_group_df = spark.sql(f"""
WITH language_summary AS (

    SELECT
        poi_group_id,

        MAX(
            CASE
                WHEN LOWER(TRIM(lang)) = 'vi' THEN 1
                ELSE 0
            END
        ) AS has_vi,

        MAX(
            CASE
                WHEN LOWER(TRIM(lang)) = 'en' THEN 1
                ELSE 0
            END
        ) AS has_en,

        MAX(
            CASE
                WHEN LOWER(TRIM(lang)) = 'ko' THEN 1
                ELSE 0
            END
        ) AS has_ko,

        CONCAT_WS(
            ',',
            SORT_ARRAY(
                COLLECT_SET(LOWER(TRIM(lang)))
            )
        ) AS existing_languages,

        COUNT(DISTINCT poi_id) AS active_poi_count

    FROM {POI_MASTER}

    WHERE poi_group_id IS NOT NULL
      AND TRIM(poi_group_id) <> ''
      AND is_active = TRUE

    GROUP BY poi_group_id
)

SELECT
    poi_group_id,
    has_vi,
    has_en,
    has_ko,
    existing_languages,
    active_poi_count,

    CONCAT_WS(
        ',',
        FILTER(
            ARRAY(
                CASE WHEN has_vi = 0 THEN 'vi' END,
                CASE WHEN has_en = 0 THEN 'en' END,
                CASE WHEN has_ko = 0 THEN 'ko' END
            ),
            x -> x IS NOT NULL
        )
    ) AS missing_languages

FROM language_summary

WHERE NOT (
       has_vi = 1
   AND has_en = 1
   AND has_ko = 1
)
""").cache()


# Materialize BEFORE any update
invalid_group_count = invalid_group_df.count()

invalid_group_df.createOrReplaceTempView(
    "v_cleanup_invalid_poi_groups"
)


# ============================================================
# 3. FREEZE ALL POI IDs BELONGING TO INVALID GROUPS
#
# IMPORTANT:
# We intentionally select ALL POIs of invalid groups,
# not only currently active rows.
#
# This gives us one fixed source for updating both tables.
# ============================================================

invalid_poi_df = spark.sql(f"""
SELECT DISTINCT
    p.poi_id,
    p.poi_group_id,
    LOWER(TRIM(p.lang)) AS lang,
    g.existing_languages,
    g.missing_languages

FROM {POI_MASTER} p

INNER JOIN v_cleanup_invalid_poi_groups g
    ON g.poi_group_id = p.poi_group_id

WHERE p.poi_id IS NOT NULL
""").cache()


# Materialize / freeze before MERGE
invalid_poi_count = invalid_poi_df.count()

invalid_poi_df.createOrReplaceTempView(
    "v_cleanup_invalid_poi_ids"
)


# ============================================================
# 4. PRE-UPDATE SUMMARY
# ============================================================

print("============================================================")
print("HISTORICAL INCOMPLETE LANGUAGE CLEANUP")
print("============================================================")
print(f"Invalid poi_group_id count : {invalid_group_count}")
print(f"POI IDs to deactivate      : {invalid_poi_count}")
print("Required languages         : vi, en, ko")
print("============================================================")


if invalid_poi_count == 0:

    print(
        "No incomplete-language POI group found. "
        "Nothing to update."
    )

else:

    # ========================================================
    # 5. DEACTIVATE POI_MASTER
    # ========================================================

    if "updated_at" in poi_master_cols:

        spark.sql(f"""
        MERGE INTO {POI_MASTER} AS tgt

        USING (
            SELECT DISTINCT poi_id
            FROM v_cleanup_invalid_poi_ids
        ) AS src

        ON tgt.poi_id = src.poi_id

        WHEN MATCHED
             AND tgt.is_active = TRUE

        THEN UPDATE SET
            tgt.is_active = FALSE,
            tgt.updated_at = CURRENT_TIMESTAMP()
        """)

    else:

        spark.sql(f"""
        MERGE INTO {POI_MASTER} AS tgt

        USING (
            SELECT DISTINCT poi_id
            FROM v_cleanup_invalid_poi_ids
        ) AS src

        ON tgt.poi_id = src.poi_id

        WHEN MATCHED
             AND tgt.is_active = TRUE

        THEN UPDATE SET
            tgt.is_active = FALSE
        """)


    # ========================================================
    # 6. DEACTIVATE AI_ENRICHED_POI
    # ========================================================

    if "updated_at" in ai_enriched_cols:

        spark.sql(f"""
        MERGE INTO {AI_ENRICHED_POI} AS tgt

        USING (
            SELECT DISTINCT poi_id
            FROM v_cleanup_invalid_poi_ids
        ) AS src

        ON tgt.poi_id = src.poi_id

        WHEN MATCHED
             AND tgt.is_active = TRUE

        THEN UPDATE SET
            tgt.is_active = FALSE,
            tgt.updated_at = CURRENT_TIMESTAMP()
        """)

    else:

        spark.sql(f"""
        MERGE INTO {AI_ENRICHED_POI} AS tgt

        USING (
            SELECT DISTINCT poi_id
            FROM v_cleanup_invalid_poi_ids
        ) AS src

        ON tgt.poi_id = src.poi_id

        WHEN MATCHED
             AND tgt.is_active = TRUE

        THEN UPDATE SET
            tgt.is_active = FALSE
        """)


    # ========================================================
    # 7. VERIFY POI_MASTER
    # ========================================================

    verify_master = spark.sql(f"""
    SELECT
        COUNT(*) AS total_detected_poi,

        SUM(
            CASE
                WHEN COALESCE(p.is_active, FALSE) = FALSE
                THEN 1
                ELSE 0
            END
        ) AS deactivated_poi,

        SUM(
            CASE
                WHEN p.is_active = TRUE
                THEN 1
                ELSE 0
            END
        ) AS still_active_poi

    FROM v_cleanup_invalid_poi_ids x

    LEFT JOIN {POI_MASTER} p
        ON p.poi_id = x.poi_id
    """).first()


    # ========================================================
    # 8. VERIFY AI_ENRICHED_POI
    # ========================================================

    verify_ai = spark.sql(f"""
    SELECT

        COUNT(DISTINCT x.poi_id) AS detected_poi,

        COUNT(DISTINCT a.poi_id) AS found_in_ai_enriched,

        COUNT(
            DISTINCT CASE
                WHEN COALESCE(a.is_active, FALSE) = FALSE
                THEN a.poi_id
            END
        ) AS ai_deactivated,

        COUNT(
            DISTINCT CASE
                WHEN a.is_active = TRUE
                THEN a.poi_id
            END
        ) AS ai_still_active

    FROM v_cleanup_invalid_poi_ids x

    LEFT JOIN {AI_ENRICHED_POI} a
        ON a.poi_id = x.poi_id
    """).first()


    print("")
    print("============================================================")
    print("POI_MASTER RESULT")
    print("============================================================")
    print(
        f"Detected POIs  : "
        f"{verify_master['total_detected_poi']}"
    )
    print(
        f"Deactivated    : "
        f"{verify_master['deactivated_poi']}"
    )
    print(
        f"Still active   : "
        f"{verify_master['still_active_poi']}"
    )


    print("")
    print("============================================================")
    print("AI_ENRICHED_POI RESULT")
    print("============================================================")
    print(
        f"Detected POIs      : "
        f"{verify_ai['detected_poi']}"
    )
    print(
        f"Found in AI table  : "
        f"{verify_ai['found_in_ai_enriched']}"
    )
    print(
        f"Deactivated        : "
        f"{verify_ai['ai_deactivated']}"
    )
    print(
        f"Still active       : "
        f"{verify_ai['ai_still_active']}"
    )


    # ========================================================
    # 9. HARD VALIDATION
    # ========================================================

    remaining_master_active = (
        verify_master["still_active_poi"] or 0
    )

    remaining_ai_active = (
        verify_ai["ai_still_active"] or 0
    )

    if remaining_master_active > 0:
        raise Exception(
            "CLEANUP_FAILED: Some invalid POIs are still active "
            f"in poi_master: {remaining_master_active}"
        )

    if remaining_ai_active > 0:
        raise Exception(
            "CLEANUP_FAILED: Some invalid POIs are still active "
            f"in ai_enriched_poi: {remaining_ai_active}"
        )

    print("")
    print("============================================================")
    print("CLEANUP SUCCESS")
    print("All detected incomplete-language POIs are inactive.")
    print("============================================================")


# ============================================================
# 10. CLEAN CACHE
# ============================================================

invalid_poi_df.unpersist()
invalid_group_df.unpersist()
```
