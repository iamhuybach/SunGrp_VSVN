```python
p_exec_id = ""
p_run_id = ""
p_pl_name = ""
p_trg_fq = ""
p_pinned_json = ""
p_mode = "RUN"
p_force = "false"
p_history_limit = "1000"
```

```python
OUT_COLS = ("media_id", "poi_id", "source_name", "photo_api_uri", "display_order", "width_px", "height_px",
            "photographer", "blob_url", "thumbnail_url", "is_cover", "is_active")


def _require(df, cols, name):
    missing = [c for c in cols if c not in df.columns]
    if missing:
        raise ConfigError(f"gld_srv_poi_media_gallery: {name} thiếu {missing}")


def build(ctx):
    media = ctx.input("media")
    _require(media, OUT_COLS, "slv_poi_media_info")
    return media.select(*OUT_COLS)


SPEC = NodeSpec(
    trg_tbl="gld_srv_poi_media_gallery",
    inputs={"media": "slv_poi_media_info"},
    pk=("media_id",),
    build=build,
    required_inputs=("media",),
    soft_delete_set={"is_active": "false"},
    write_mode="MERGE",
)
```

```python
RES = run_node(SPEC, p_exec_id, p_run_id, p_pinned_json, p_mode, p_force, p_pl_name, p_trg_fq, p_history_limit)
```
