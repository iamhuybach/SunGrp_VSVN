```python
# BEGIN_AMENITY_SEED
SEED_SECTORS = ("accommodation", "f_n_b", "attraction", "shopping_n_retail")
SEED_GROUPS = (
    "general_amenities",
    "accessibility",
    "specialty_tags",
    "space_and_services",
    "facilities_and_services",
)
SEED_SKIP = ("Bổ sung sau", "Fallback", "Không có field", "Không có field riêng")
SEED_ROWS = [('google',
  'accommodation',
  'general_amenities',
  'parkingOptions.freeStreetParking = true',
  'Free Street Parking',
  'Đỗ xe trên phố'),
 ('google',
  'accommodation',
  'general_amenities',
  'parkingOptions.valetParking = true',
  'Valet Parking',
  'Dịch vụ Valet'),
 ('google',
  'accommodation',
  'general_amenities',
  'parkingOptions.freeGarageParking = true',
  'Free Garage Parking',
  'Bãi xe trong nhà (Miễn phí)'),
 ('google',
  'accommodation',
  'general_amenities',
  'parkingOptions.paidGarageParking = true',
  'Paid Garage Parking',
  'Bãi xe trong nhà (Thu phí)'),
 ('google', 'accommodation', 'general_amenities', 'outdoorSeating = true', 'Outdoor Seating', 'Không gian ngoài trời'),
 ('google', 'accommodation', 'general_amenities', 'restroom = true', 'Restroom', 'Nhà vệ sinh'),
 ('google', 'accommodation', 'general_amenities', 'goodForChildren = true', 'Good for Children', 'Thân thiện trẻ em'),
 ('google', 'accommodation', 'general_amenities', 'goodForGroups = true', 'Good for Groups', 'Phù hợp đoàn nhóm'),
 ('google', 'accommodation', 'general_amenities', 'liveMusic = true', 'Live Music', 'Nhạc Live'),
 ('partner', 'accommodation', 'general_amenities', 'Hồ bơi', 'Swimming Pool', 'Hồ bơi'),
 ('google', 'accommodation', 'general_amenities', 'types[] = swimming_pool', 'Swimming Pool', 'Hồ bơi'),
 ('partner', 'accommodation', 'general_amenities', 'Wi-Fi miễn phí', 'Free Wi-Fi', 'Wi-Fi miễn phí'),
 ('partner', 'accommodation', 'general_amenities', 'Bãi đậu xe', 'Parking', 'Bãi đậu xe'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.freeParkingLot = true', 'Parking', 'Bãi đậu xe'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.paidParkingLot = true', 'Parking', 'Bãi đậu xe'),
 ('partner', 'accommodation', 'general_amenities', 'Xe đưa đón sân bay', 'Airport Shuttle', 'Xe đưa đón sân bay'),
 ('partner', 'accommodation', 'general_amenities', 'Nhà hàng', 'Restaurant', 'Nhà hàng'),
 ('google',
  'accommodation',
  'general_amenities',
  'types[] có enum thuộc business_sector F&B',
  'Restaurant',
  'Nhà hàng'),
 ('google', 'accommodation', 'general_amenities', 'serves_meals = true', 'Restaurant', 'Nhà hàng'),
 ('partner', 'accommodation', 'general_amenities', 'Lễ tân 24h', '24-hour Front Desk', 'Lễ tân 24h'),
 ('partner', 'accommodation', 'general_amenities', 'Quầy bar', 'Bar', 'Quầy bar'),
 ('google',
  'accommodation',
  'general_amenities',
  'types[] có enum thuộc business_category = Bar & Pub/Nightlife servesBeer = true',
  'Bar',
  'Quầy bar'),
 ('google', 'accommodation', 'general_amenities', 'servesCocktails = true', 'Bar', 'Quầy bar'),
 ('partner', 'accommodation', 'general_amenities', 'Spa & Wellness', 'Spa & Wellness', 'Spa & Wellness'),
 ('google',
  'accommodation',
  'general_amenities',
  'types[] có enum thuộc business_category = Day Spa/Therapeutic Massage/Clinic-Aesthetic Spa/Hair Salon/Nail-Lash '
  'Salon/Beauty Salon/Barbershop',
  'Spa & Wellness',
  'Spa & Wellness'),
 ('partner', 'accommodation', 'general_amenities', 'Phòng Gym', 'Gym', 'Phòng Gym'),
 ('google',
  'accommodation',
  'general_amenities',
  'types[] có enum thuộc business_category = Gym-Fitness/Yoga-Pilates Studio',
  'Gym',
  'Phòng Gym'),
 ('partner', 'accommodation', 'general_amenities', 'Xông hơi', 'Sauna', 'Xông hơi'),
 ('google', 'accommodation', 'general_amenities', 'types[] = hot_spring, sauna', 'Sauna', 'Xông hơi'),
 ('partner', 'accommodation', 'general_amenities', 'Bồn tắm nóng (Jacuzzi)', 'Jacuzzi', 'Bồn tắm nóng'),
 ('partner', 'accommodation', 'general_amenities', 'Sân vườn', 'Garden', 'Sân vườn'),
 ('google', 'accommodation', 'general_amenities', 'types[] = garden, botanical_garden', 'Garden', 'Sân vườn'),
 ('partner', 'accommodation', 'general_amenities', 'Phòng họp', 'Meeting Rooms', 'Phòng họp'),
 ('google', 'accommodation', 'general_amenities', 'types[] = meeting_rooms', 'Meeting Rooms', 'Phòng họp'),
 ('partner', 'accommodation', 'general_amenities', 'Khu vực làm việc chung', 'Co-working Space', 'Khu làm việc chung'),
 ('partner', 'accommodation', 'general_amenities', 'Giữ hành lý', 'Luggage Storage', 'Giữ hành lý'),
 ('partner', 'accommodation', 'general_amenities', 'Thu đổi ngoại tệ', 'Currency Exchange', 'Thu đổi ngoại tệ'),
 ('partner', 'accommodation', 'general_amenities', 'Trạm sạc xe điện', 'EV Charging Station', 'Trạm sạc xe điện'),
 ('google',
  'accommodation',
  'general_amenities',
  'parkingOptions.freeElectricVehicleParking = true',
  'EV Charging Station',
  'Trạm sạc xe điện'),
 ('google',
  'accommodation',
  'general_amenities',
  'parkingOptions.paidElectricVehicleParking = true',
  'EV Charging Station',
  'Trạm sạc xe điện'),
 ('google',
  'accommodation',
  'general_amenities',
  'types[] có electric_vehicle_charging_station',
  'EV Charging Station',
  'Trạm sạc xe điện'),
 ('google',
  'accommodation',
  'accessibility',
  'accessibilityOptions.wheelchairAccessibleSeating = true',
  'Wheelchair Accessible Seating',
  'Chỗ ngồi cho người khuyết tật'),
 ('google',
  'accommodation',
  'accessibility',
  'accessibilityOptions.wheelchairAccessibleElevator = true',
  'Wheelchair Accessible Elevator',
  'Thang máy hỗ trợ xe lăn'),
 ('partner', 'accommodation', 'accessibility', 'Lối đi xe lăn', 'Wheelchair Accessible Entrance', 'Lối đi xe lăn'),
 ('google',
  'accommodation',
  'accessibility',
  'accessibilityOptions.wheelchairAccessibleEntrance = true',
  'Wheelchair Accessible Entrance',
  'Lối đi xe lăn'),
 ('partner', 'accommodation', 'accessibility', 'Thang máy có chữ nổi', 'Braille Elevator', 'Thang máy có chữ nổi'),
 ('partner',
  'accommodation',
  'accessibility',
  'Nhà vệ sinh có tay vịn',
  'Wheelchair Accessible Restroom',
  'Nhà vệ sinh có tay vịn'),
 ('google',
  'accommodation',
  'accessibility',
  'accessibilityOptions.wheelchairAccessibleRestroom = true',
  'Wheelchair Accessible Restroom',
  'Nhà vệ sinh có tay vịn'),
 ('partner', 'accommodation', 'accessibility', 'Bồn rửa mặt thấp', 'Lowered Sink', 'Bồn rửa mặt thấp'),
 ('partner',
  'accommodation',
  'accessibility',
  'Bãi đỗ xe cho người khuyết tật',
  'Accessible Parking',
  'Bãi đỗ xe cho người khuyết tật'),
 ('google',
  'accommodation',
  'accessibility',
  'accessibilityOptions.wheelchairAccessibleParking = true',
  'Accessible Parking',
  'Bãi đỗ xe cho người khuyết tật'),
 ('google', 'f_n_b', 'specialty_tags', 'dim_sum_restaurant', 'Dim Sum', 'Dimsum'),
 ('google', 'f_n_b', 'specialty_tags', 'sushi_restaurant', 'Sushi', 'Sushi'),
 ('google', 'f_n_b', 'specialty_tags', 'dumpling_restaurant', 'Dumpling', 'Há cảo – Bánh bao'),
 ('google', 'f_n_b', 'specialty_tags', 'tonkatsu_restaurant', 'Tonkatsu', 'Tonkatsu'),
 ('google', 'f_n_b', 'specialty_tags', 'japanese_curry_restaurant', 'Japanese Curry', 'Cà ri Nhật'),
 ('google', 'f_n_b', 'specialty_tags', 'cantonese_restaurant', 'Cantonese', 'Ẩm thực Quảng Đông'),
 ('google', 'f_n_b', 'specialty_tags', 'chinese_noodle_restaurant', 'Chinese Noodle', 'Mì Trung Quốc'),
 ('google', 'f_n_b', 'specialty_tags', 'ramen_restaurant', 'Ramen', 'Ramen'),
 ('google',
  'f_n_b',
  'specialty_tags',
  'yakiniku_restaurant, yakitori_restaurant',
  'Yakiniku / Teppanyaki',
  'Yakiniku / Teppanyaki'),
 ('google', 'f_n_b', 'specialty_tags', 'japanese_izakaya_restaurant', 'Izakaya', 'Izakaya'),
 ('google', 'f_n_b', 'specialty_tags', 'korean_barbecue_restaurant', 'Korean BBQ', 'BBQ Hàn Quốc'),
 ('google', 'f_n_b', 'specialty_tags', 'pizza_restaurant', 'Pizza', 'Pizza'),
 ('google', 'f_n_b', 'specialty_tags', 'hamburger_restaurant', 'Hamburger', 'Hamburger'),
 ('google', 'f_n_b', 'specialty_tags', 'brunch_restaurant', 'Brunch', 'Brunch'),
 ('google', 'f_n_b', 'specialty_tags', 'cat_cafe, dog_cafe', 'Pet Café', 'Quán cà phê thú cưng'),
 ('google', 'f_n_b', 'specialty_tags', 'tapas_restaurant', 'Tapas', 'Tapas'),
 ('google', 'f_n_b', 'specialty_tags', 'soup_restaurant', 'Soup', 'Súp'),
 ('partner', 'f_n_b', 'specialty_tags', 'Street Food', 'Street Food', 'Ẩm thực đường phố'),
 ('google',
  'f_n_b',
  'specialty_tags',
  'types[] = food_stand, street_food_gathering',
  'Street Food',
  'Ẩm thực đường phố'),
 ('partner', 'f_n_b', 'specialty_tags', 'Seafood', 'Seafood', 'Hải sản'),
 ('google', 'f_n_b', 'specialty_tags', 'types[] = seafood_restaurant, oyster_bar_restaurant', 'Seafood', 'Hải sản'),
 ('partner', 'f_n_b', 'specialty_tags', 'Steakhouse', 'Steakhouse', 'Bít tết'),
 ('google', 'f_n_b', 'specialty_tags', 'types[] = steakhouse', 'Steakhouse', 'Bít tết'),
 ('partner', 'f_n_b', 'specialty_tags', 'Office Lunch', 'Office Lunch', 'Cơm văn phòng'),
 ('partner', 'f_n_b', 'specialty_tags', 'Milk Tea', 'Milk Tea', 'Trà sữa'),
 ('partner', 'f_n_b', 'specialty_tags', 'Bakery', 'Bakery', 'Tiệm bánh'),
 ('google', 'f_n_b', 'specialty_tags', 'types[] = bakery, pastry_shop, cake_shop', 'Bakery', 'Tiệm bánh'),
 ('partner', 'f_n_b', 'specialty_tags', 'Bar-Pub', 'Bar & Pub', 'Bar – Pub'),
 ('google',
  'f_n_b',
  'specialty_tags',
  'types[] thuộc business_category = Bar & Pub/Nightlife',
  'Bar & Pub',
  'Bar – Pub'),
 ('partner', 'f_n_b', 'specialty_tags', 'Rooftop Bar', 'Rooftop Bar', 'Bar sân thượng'),
 ('google', 'f_n_b', 'space_and_services', 'restroom = true', 'Restroom', 'Nhà vệ sinh'),
 ('google', 'f_n_b', 'space_and_services', 'goodForChildren = true', 'Good for Children', 'Thân thiện trẻ em'),
 ('partner', 'f_n_b', 'space_and_services', 'Outdoor Seating', 'Outdoor Seating', 'Chỗ ngồi ngoài trời'),
 ('google', 'f_n_b', 'space_and_services', 'outdoorSeating = true = true', 'Outdoor Seating', 'Chỗ ngồi ngoài trời'),
 ('partner', 'f_n_b', 'space_and_services', 'Private Room (VIP)', 'Private Room (VIP)', 'Phòng riêng (VIP)'),
 ('partner', 'f_n_b', 'space_and_services', 'Scenic View', 'Scenic View', 'Góc nhìn đẹp'),
 ('partner', 'f_n_b', 'space_and_services', 'Romantic Decor', 'Romantic Decor', 'Trang trí lãng mạn'),
 ('partner', 'f_n_b', 'space_and_services', 'High Chair', 'High Chair', 'Có ghế trẻ em'),
 ('partner', 'f_n_b', 'space_and_services', 'Wheelchair Accessible', 'Wheelchair Accessible', 'Lối đi xe lăn'),
 ('google',
  'f_n_b',
  'space_and_services',
  'accessibilityOptions.wheelchairAccessibleEntrance = true',
  'Wheelchair Accessible',
  'Lối đi xe lăn'),
 ('partner', 'f_n_b', 'space_and_services', 'Sports Screening', 'Sports Screening', 'Chiếu thể thao trực tiếp'),
 ('google',
  'f_n_b',
  'space_and_services',
  'goodForWatchingSports = true',
  'Sports Screening',
  'Chiếu thể thao trực tiếp'),
 ('partner', 'f_n_b', 'space_and_services', 'Live Music', 'Live Music', 'Nhạc sống'),
 ('google', 'f_n_b', 'space_and_services', 'liveMusic = true', 'Live Music', 'Nhạc sống'),
 ('partner', 'f_n_b', 'space_and_services', 'Car Parking', 'Car Parking', 'Bãi đỗ xe ô tô'),
 ('google',
  'f_n_b',
  'space_and_services',
  'parkingOptions.freeParkingLot, parkingOptions.paidParkingLot = true',
  'Car Parking',
  'Bãi đỗ xe ô tô'),
 ('partner', 'f_n_b', 'space_and_services', 'Table Service', 'Table Service', 'Phục vụ tại bàn'),
 ('google', 'f_n_b', 'space_and_services', 'dineIn = true', 'Table Service', 'Phục vụ tại bàn'),
 ('google',
  'attraction',
  'facilities_and_services',
  'accessibilityOptions.wheelchairAccessibleEntrance',
  'Wheelchair Accessible Entrance',
  'Lối đi xe lăn'),
 ('google',
  'attraction',
  'facilities_and_services',
  'accessibilityOptions.wheelchairAccessibleParking',
  'Accessible Parking',
  'Bãi đỗ xe người khuyết tật'),
 ('google',
  'attraction',
  'facilities_and_services',
  'accessibilityOptions.wheelchairAccessibleRestroom',
  'Wheelchair Accessible Restroom',
  'Nhà vệ sinh xe lăn'),
 ('google', 'attraction', 'facilities_and_services', 'restroom', 'Restroom', 'Nhà vệ sinh'),
 ('google', 'attraction', 'facilities_and_services', 'goodForChildren', 'Good for Children', 'Thân thiện trẻ em'),
 ('google', 'attraction', 'facilities_and_services', 'liveMusic', 'Live Music', 'Nhạc Live'),
 ('partner', 'attraction', 'facilities_and_services', 'Indoor', 'Indoor', 'Trong nhà'),
 ('partner', 'attraction', 'facilities_and_services', 'Outdoor', 'Outdoor', 'Ngoài trời'),
 ('partner', 'attraction', 'facilities_and_services', 'Covered', 'Covered', 'Có mái che'),
 ('partner', 'attraction', 'facilities_and_services', 'Family Friendly', 'Family Friendly', 'Thân thiện với gia đình'),
 ('google', 'attraction', 'facilities_and_services', 'goodForChildren', 'Family Friendly', 'Thân thiện với gia đình'),
 ('partner', 'attraction', 'facilities_and_services', 'Couples', 'Couples', 'Cặp đôi'),
 ('partner', 'attraction', 'facilities_and_services', 'Group', 'Group', 'Nhóm bạn'),
 ('google', 'attraction', 'facilities_and_services', 'goodForGroups', 'Group', 'Nhóm bạn'),
 ('partner', 'attraction', 'facilities_and_services', 'History', 'History', 'Lịch sử'),
 ('google',
  'attraction',
  'facilities_and_services',
  'types[] có enum thuộc business_category = Historical Site',
  'History',
  'Lịch sử'),
 ('partner', 'attraction', 'facilities_and_services', 'Art', 'Art', 'Nghệ thuật'),
 ('google', 'attraction', 'facilities_and_services', 'types[] có art_gallery', 'Art', 'Nghệ thuật'),
 ('partner', 'attraction', 'facilities_and_services', 'Science', 'Science', 'Khoa học'),
 ('google', 'attraction', 'facilities_and_services', 'types[] có planetarium', 'Science', 'Khoa học'),
 ('partner', 'attraction', 'facilities_and_services', 'Interactive', 'Interactive', 'Tương tác'),
 ('partner', 'attraction', 'facilities_and_services', 'Educational', 'Educational', 'Giáo dục'),
 ('partner', 'attraction', 'facilities_and_services', 'Spiritual', 'Spiritual', 'Tâm linh'),
 ('google',
  'attraction',
  'facilities_and_services',
  'types[] có business_category = Religious Site',
  'Spiritual',
  'Tâm linh'),
 ('partner', 'attraction', 'facilities_and_services', 'Parking', 'Parking', 'Bãi đậu xe'),
 ('google', 'attraction', 'facilities_and_services', 'parkingOptions.freeParkingLot = true', 'Parking', 'Bãi đậu xe'),
 ('google', 'attraction', 'facilities_and_services', 'parkingOptions.paidParkingLot = true', 'Parking', 'Bãi đậu xe'),
 ('partner', 'attraction', 'facilities_and_services', 'Free Wi-Fi', 'Free Wi-Fi', 'Wi-Fi miễn phí'),
 ('partner', 'attraction', 'facilities_and_services', 'Lockers', 'Lockers', 'Tủ gửi đồ'),
 ('partner', 'attraction', 'facilities_and_services', 'Restaurant-Cafe', 'Restaurant-Cafe', 'Nhà hàng – Quán ăn'),
 ('google',
  'attraction',
  'facilities_and_services',
  'types[] có restaurant, cafe, coffee_shop',
  'Restaurant-Cafe',
  'Nhà hàng – Quán ăn'),
 ('partner', 'attraction', 'facilities_and_services', 'Gift Shop', 'Gift Shop', 'Cửa hàng lưu niệm'),
 ('google', 'attraction', 'facilities_and_services', 'types[] có gift_shop', 'Gift Shop', 'Cửa hàng lưu niệm'),
 ('partner', 'attraction', 'facilities_and_services', 'Stroller Rental', 'Stroller Rental', 'Cho thuê xe đẩy trẻ em'),
 ('partner', 'attraction', 'facilities_and_services', 'Wheelchair Rental', 'Wheelchair Rental', 'Cho thuê xe lăn'),
 ('partner', 'attraction', 'facilities_and_services', 'First Aid Room', 'First Aid Room', 'Phòng y tế'),
 ('partner',
  'attraction',
  'facilities_and_services',
  'On-site Tour Guide',
  'On-site Tour Guide',
  'Hướng dẫn viên tại điểm'),
 ('partner', 'attraction', 'facilities_and_services', 'Audio Guide', 'Audio Guide', 'Máy hướng dẫn tự động'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Tax Refund', 'Tax Refund', 'Hoàn thuế'),
 ('partner',
  'shopping_n_retail',
  'facilities_and_services',
  'Shipping-Delivery',
  'Shipping-Delivery',
  'Giao hàng tận nơi'),
 ('google', 'shopping_n_retail', 'facilities_and_services', 'delivery', 'Shipping-Delivery', 'Giao hàng tận nơi'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Gift Wrapping', 'Gift Wrapping', 'Gói quà miễn phí'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Fitting Room', 'Fitting Room', 'Phòng thử đồ'),
 ('partner',
  'shopping_n_retail',
  'facilities_and_services',
  'Personalization-Engraving',
  'Personalization-Engraving',
  'Khắc tên – Cá nhân hóa'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Tailoring', 'Tailoring', 'May đo theo yêu cầu'),
 ('partner',
  'shopping_n_retail',
  'facilities_and_services',
  'Credit Card (Visa/Master)',
  'Credit Card',
  'Chấp nhận thẻ tín dụng'),
 ('google',
  'shopping_n_retail',
  'facilities_and_services',
  'paymentOptions.acceptsCreditCards',
  'Credit Card',
  'Chấp nhận thẻ tín dụng'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'E-Wallet', 'E-Wallet', 'Ví điện tử'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'VAT Invoice', 'VAT Invoice', 'Xuất hóa đơn VAT'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Flexible Return', 'Flexible Return', 'Đổi trả linh hoạt'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Parking', 'Parking', 'Bãi đỗ xe'),
 ('google',
  'shopping_n_retail',
  'facilities_and_services',
  'parkingOptions.freeParkingLot, parkingOptions.paidParkingLot',
  'Parking',
  'Bãi đỗ xe'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Free Wi-Fi', 'Free Wi-Fi', 'Wi-Fi miễn phí')]


def validate_seed(rows):
    """Seed phải đúng nguồn, sector, nhóm và khoá 5 cột."""
    keys = [row[:5] for row in rows]
    if len(keys) != len(set(keys)):
        raise ValueError("ref_dictionary_amenities: trùng khoá 5 cột")
    for source, sector, group, raw, amenity, _vi in rows:
        if source not in {"google", "partner"}:
            raise ValueError(f"source lạ: {source}")
        if sector not in SEED_SECTORS:
            raise ValueError(f"sector lạ: {sector}")
        if group not in SEED_GROUPS:
            raise ValueError(f"group lạ: {group}")
        if not raw or raw in SEED_SKIP:
            raise ValueError(f"amenity_raw không hợp lệ: {raw}")
        if not amenity:
            raise ValueError(f"amenity trống: {raw}")
# END_AMENITY_SEED
```

```python
spark.sql("""
CREATE TABLE IF NOT EXISTS lh_vv_silver.dbo.ref_dictionary_amenities (
    source_name     STRING NOT NULL,
    business_sector STRING NOT NULL,
    amenity_group   STRING NOT NULL,
    amenity_raw     STRING NOT NULL,
    amenity         STRING NOT NULL,
    amenity_vi      STRING,
    is_active       BOOLEAN,
    updated_at      TIMESTAMP
) USING DELTA
TBLPROPERTIES ('delta.autoOptimize.optimizeWrite' = 'true', 'delta.autoOptimize.autoCompact' = 'true')
""")
spark.sql("""
CREATE TABLE IF NOT EXISTS lh_vv_silver.dbo.slv_poi_amenity (
    poi_id          STRING NOT NULL,
    business_sector STRING NOT NULL,
    amenity_group   STRING NOT NULL,
    amenity_raw     STRING NOT NULL,
    amenity         STRING,
    amenity_vi      STRING,
    row_hash        STRING,
    created_at      TIMESTAMP,
    updated_at      TIMESTAMP,
    deleted_at      TIMESTAMP
) USING DELTA
TBLPROPERTIES ('delta.autoOptimize.optimizeWrite' = 'true', 'delta.autoOptimize.autoCompact' = 'true')
""")
EXPECTED = {
    REF_FQ: [
        ("source_name", "string", False),
        ("business_sector", "string", False),
        ("amenity_group", "string", False),
        ("amenity_raw", "string", False),
        ("amenity", "string", False),
        ("amenity_vi", "string", True),
        ("is_active", "boolean", True),
        ("updated_at", "timestamp", True),
    ],
    AMENITY_FQ: [
        ("poi_id", "string", False),
        ("business_sector", "string", False),
        ("amenity_group", "string", False),
        ("amenity_raw", "string", False),
        ("amenity", "string", True),
        ("amenity_vi", "string", True),
        ("row_hash", "string", True),
        ("created_at", "timestamp", True),
        ("updated_at", "timestamp", True),
        ("deleted_at", "timestamp", True),
    ],
}
for fq, expected in EXPECTED.items():
    got = [(field.name.lower(), field.dataType.simpleString(), field.nullable) for field in spark.table(fq).schema.fields]
    if got != expected:
        raise ConfigError(f"{fq} lệch cột hoặc nullability: có {got}, cần {expected}")
```

```python
def _seed_merge():
    spark.sql("""
        MERGE INTO lh_vv_silver.dbo.ref_dictionary_amenities AS t
        USING seed_amenities AS s
        ON t.source_name = s.source_name
       AND t.business_sector = s.business_sector
       AND t.amenity_group = s.amenity_group
       AND t.amenity_raw = s.amenity_raw
       AND t.amenity = s.amenity
        WHEN NOT MATCHED THEN INSERT (
            source_name, business_sector, amenity_group, amenity_raw, amenity, amenity_vi, is_active, updated_at
        ) VALUES (
            s.source_name, s.business_sector, s.amenity_group, s.amenity_raw, s.amenity, s.amenity_vi,
            true, current_timestamp()
        )
    """)


missing_before = spark.sql("""
    SELECT count(*) AS n
    FROM seed_amenities s
    LEFT ANTI JOIN lh_vv_silver.dbo.ref_dictionary_amenities t
      ON t.source_name = s.source_name
     AND t.business_sector = s.business_sector
     AND t.amenity_group = s.amenity_group
     AND t.amenity_raw = s.amenity_raw
     AND t.amenity = s.amenity
""").first()["n"]
with_retry(_seed_merge, "seed ref_dictionary_amenities")
INSERTED = int(missing_before)

dups = spark.sql("""
    SELECT count(*) AS n FROM (
        SELECT source_name, business_sector, amenity_group, amenity_raw, amenity
        FROM lh_vv_silver.dbo.ref_dictionary_amenities
        GROUP BY source_name, business_sector, amenity_group, amenity_raw, amenity
        HAVING count(*) > 1
    )
""").first()["n"]
if dups:
    raise ConfigError(f"ref_dictionary_amenities: {dups} khoá 5 cột trùng")

missing_after = spark.sql("""
    SELECT count(*) AS n
    FROM seed_amenities s
    LEFT ANTI JOIN lh_vv_silver.dbo.ref_dictionary_amenities t
      ON t.source_name = s.source_name
     AND t.business_sector = s.business_sector
     AND t.amenity_group = s.amenity_group
     AND t.amenity_raw = s.amenity_raw
     AND t.amenity = s.amenity
""").first()["n"]
if missing_after:
    raise ConfigError(f"ref_dictionary_amenities: thiếu {missing_after} khoá seed")

DRIFT = int(spark.sql("""
    SELECT count(*) AS n
    FROM seed_amenities s
    JOIN lh_vv_silver.dbo.ref_dictionary_amenities t
      ON t.source_name = s.source_name
     AND t.business_sector = s.business_sector
     AND t.amenity_group = s.amenity_group
     AND t.amenity_raw = s.amenity_raw
     AND t.amenity = s.amenity
    WHERE NOT (t.amenity_vi <=> s.amenity_vi AND t.is_active = true)
""").first()["n"])
print(f"inserted={INSERTED} drift={DRIFT}")
```

```python
import uuid

def _write_edges():
    """Ghi cạnh UPSERT và RETIRE trong một MERGE. Trùng khoá hoặc owner khác thì fail."""
    if entity_edges_ok != 5:
        raise ConfigError("slv_poi_amenity: chưa có cạnh slv_poi_entity_non_enrich, chạy NB_SETUP_POI_ENTITY trước")
    cfg = "lh_vv_ctrl.dbo.ctrl_mng_pipeline_config"
    has_nb = "nb_name" in table_columns(cfg)
    nb_match = " AND t.nb_name <=> CAST(NULL AS STRING)" if has_nb else ""
    nb_set = ", nb_name = NULL" if has_nb else ""
    nb_col = ", nb_name" if has_nb else ""
    nb_val = ", CAST(NULL AS STRING)" if has_nb else ""
    nb_ok = " AND nb_name IS NULL" if has_nb else ""
    dup = spark.sql(f"""
        SELECT trg_tbl, src_tbl, count(*) AS n
        FROM {cfg}
        WHERE pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
          AND (
            (trg_tbl = 'slv_poi_amenity' AND src_tbl = 'slv_3p_poi_amenity')
            OR (trg_tbl = 'slv_poi_amenity' AND src_tbl = 'slv_poi_entity_non_enrich')
            OR (trg_tbl = 'slv_poi_amenity' AND src_tbl = 'ref_dictionary_amenities')
            OR (trg_tbl = 'slv_poi_amenity' AND src_tbl = 'slv_poi_entity')
          )
        GROUP BY trg_tbl, src_tbl
        HAVING count(*) > 1
    """).collect()
    if dup:
        names = ", ".join(f"{row.trg_tbl}/{row.src_tbl}={row.n}" for row in dup)
        raise ConfigError(f"slv_poi_amenity: trùng cạnh, không ghi đè: {names}")
    owners = spark.sql(f"""
        SELECT DISTINCT pl_name
        FROM {cfg}
        WHERE trg_schema = 'lh_vv_silver.dbo'
          AND trg_tbl = 'slv_poi_amenity'
          AND is_active = 1
          AND pl_name <> 'PL_VV_TRANS_SLV_TO_SLV_1H'
    """).collect()
    if owners:
        names = ", ".join(row.pl_name for row in owners)
        raise ConfigError(f"slv_poi_amenity đã thuộc pl_name khác: {names}")
    missing = spark.sql(f"""
        SELECT count(*) AS n
        FROM VALUES (1, 'slv_3p_poi_amenity'), (2, 'slv_poi_entity_non_enrich'), (3, 'ref_dictionary_amenities') AS t(seq, src_tbl)
        WHERE NOT EXISTS (
            SELECT 1 FROM {cfg} c
            WHERE c.pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
              AND c.src_tbl = t.src_tbl AND c.trg_tbl = 'slv_poi_amenity')
    """).first()["n"]
    retired = spark.sql(f"""
        SELECT count(*) AS n FROM {cfg}
        WHERE pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
          AND (
            (trg_tbl = 'slv_poi_amenity' AND src_tbl = 'slv_poi_entity' AND NOT (is_active <=> 0))
          )
    """).first()["n"]
    spark.sql(f"""
        MERGE INTO {cfg} AS t
        USING (
            SELECT seq, src_tbl, trg_tbl, action
            FROM VALUES
            (1, 'slv_3p_poi_amenity', 'slv_poi_amenity', 'UPSERT'),
            (2, 'slv_poi_entity_non_enrich', 'slv_poi_amenity', 'UPSERT'),
            (3, 'ref_dictionary_amenities', 'slv_poi_amenity', 'UPSERT'),
            (0, 'slv_poi_entity', 'slv_poi_amenity', 'RETIRE') AS v(seq, src_tbl, trg_tbl, action)
        ) AS s
        ON t.pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
           AND t.src_tbl = s.src_tbl
           AND t.trg_tbl = s.trg_tbl
        WHEN MATCHED AND s.action = 'RETIRE' AND NOT (t.is_active <=> 0) THEN UPDATE SET
            is_active = 0
        WHEN MATCHED AND s.action = 'UPSERT' AND NOT (
            t.is_active <=> 1 AND t.priority <=> 1 AND t.load_mode <=> 'RECOMPUTE'
            AND t.src_schema <=> 'lh_vv_silver.dbo' AND t.trg_schema <=> 'lh_vv_silver.dbo'
            AND t.align_path <=> CAST(NULL AS STRING) AND t.dedup_order <=> CAST(NULL AS STRING){nb_match}
        ) THEN UPDATE SET
            src_schema = 'lh_vv_silver.dbo',
            trg_schema = 'lh_vv_silver.dbo',
            is_active = 1,
            priority = 1,
            load_mode = 'RECOMPUTE',
            align_path = NULL,
            dedup_order = NULL{nb_set}
        WHEN NOT MATCHED AND s.action = 'UPSERT' THEN INSERT (
            id, pl_name, src_schema, src_tbl, trg_schema, trg_tbl,
            is_active, priority, load_mode, align_path, dedup_order{nb_col}
        ) VALUES (
            COALESCE((SELECT MAX(id) FROM {cfg}), 0) + s.seq,
            'PL_VV_TRANS_SLV_TO_SLV_1H', 'lh_vv_silver.dbo', s.src_tbl, 'lh_vv_silver.dbo', s.trg_tbl,
            1, 1, 'RECOMPUTE', CAST(NULL AS STRING), CAST(NULL AS STRING){nb_val}
        )
    """)
    ok = spark.sql(f"""
        SELECT src_tbl
        FROM {cfg}
        WHERE pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
          AND src_schema = 'lh_vv_silver.dbo'
          AND trg_schema = 'lh_vv_silver.dbo'
          AND trg_tbl = 'slv_poi_amenity'
          AND is_active = 1 AND priority = 1 AND load_mode = 'RECOMPUTE'
          AND align_path IS NULL AND dedup_order IS NULL{nb_ok}
          AND src_tbl IN ('slv_3p_poi_amenity', 'slv_poi_entity_non_enrich', 'ref_dictionary_amenities')
    """).collect()
    counts = {}
    for row in ok:
        counts[row.src_tbl] = counts.get(row.src_tbl, 0) + 1
    if counts != {"slv_3p_poi_amenity": 1, "slv_poi_entity_non_enrich": 1, "ref_dictionary_amenities": 1}:
        raise ConfigError(f"slv_poi_amenity: cạnh active không đúng {counts}")
    extra = spark.sql(f"""
        SELECT pl_name, src_tbl
        FROM {cfg}
        WHERE trg_schema = 'lh_vv_silver.dbo' AND trg_tbl = 'slv_poi_amenity' AND is_active = 1
          AND NOT (
              pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
              AND src_schema = 'lh_vv_silver.dbo'
              AND src_tbl IN ('slv_3p_poi_amenity', 'slv_poi_entity_non_enrich', 'ref_dictionary_amenities')
              AND priority = 1 AND load_mode = 'RECOMPUTE'
              AND align_path IS NULL AND dedup_order IS NULL{nb_ok}
          )
    """).collect()
    if extra:
        names = ", ".join(f"{row.pl_name}/{row.src_tbl}" for row in extra)
        raise ConfigError(f"slv_poi_amenity: còn cạnh active ngoài cạnh chuẩn: {names}")
    
    stale = spark.sql(f"""
        SELECT count(*) AS n FROM {cfg}
        WHERE trg_tbl = 'slv_poi_amenity' AND src_tbl = 'slv_poi_entity' AND is_active = 1
    """).first()["n"]
    if stale:
        raise ConfigError(f"slv_poi_amenity: còn cạnh active slv_poi_entity")
    return int(missing), int(retired)


PL = "PL_VV_TRANS_SLV_TO_SLV_1H"
SETUP_EXEC = str(uuid.uuid4())
ensure_lock_row(PL, current_notebook_name("NB_SETUP_POI_AMENITY"))
ok, who = acquire_flow_lock(PL, SETUP_EXEC, SETUP_EXEC, LOCK_TIMEOUT_DEFAULT_MIN)
if not ok:
    raise ConfigError(f"Luồng {PL} đang chạy ({who}); chạy setup sau")
status = "FAILED"
try:
    EDGES_INSERTED, EDGES_RETIRED = with_retry(_write_edges, "edges slv_poi_amenity")
    status = "SUCCESS"
    print(f"edges_inserted={EDGES_INSERTED} edges_retired={EDGES_RETIRED}")
finally:
    release_flow_lock(PL, SETUP_EXEC, status)
```
