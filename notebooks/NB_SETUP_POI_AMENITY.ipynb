```python
# BEGIN_AMENITY_SEED
import re

AMENITY_FLAG_RE = re.compile(r"([A-Za-z][A-Za-z0-9]*(?:\.[A-Za-z][A-Za-z0-9]*)?) = true")
AMENITY_TYPES_RE = re.compile(r"types\[\] = (.+)")
# Khoá 4 cột trùng có sẵn từ mapping, dạng giữ nguyên, không phải key khớp.
# Câu types[] có/thuộc -> json_type. place_type google trong ref_dictionary_place_type.
# Day Spa không có category đúng tên: dùng category Spa.
# electric_vehicle_charging_station không có trong từ điển: giữ token.
AMENITY_JSON_TYPE = {
    'types[] có enum thuộc business_sector F&B': "json_type in ('vietnamese_restaurant', 'chinese_restaurant', 'japanese_restaurant', 'korean_restaurant', 'thai_restaurant', 'french_restaurant', 'italian_restaurant', 'pizza_restaurant', 'american_restaurant', 'european_restaurant', 'british_restaurant', 'german_restaurant', 'spanish_restaurant', 'greek_restaurant', 'steak_house', 'portuguese_restaurant', 'austrian_restaurant', 'bavarian_restaurant', 'belgian_restaurant', 'danish_restaurant', 'dutch_restaurant', 'scandinavian_restaurant', 'swiss_restaurant', 'polish_restaurant', 'romanian_restaurant', 'russian_restaurant', 'ukrainian_restaurant', 'croatian_restaurant', 'czech_restaurant', 'hungarian_restaurant', 'australian_restaurant', 'californian_restaurant', 'tex_mex_restaurant', 'fish_and_chips_restaurant', 'basque_restaurant', 'irish_restaurant', 'soul_food_restaurant', 'southwestern_us_restaurant', 'eastern_european_restaurant', 'indian_restaurant', 'north_indian_restaurant', 'south_indian_restaurant', 'indonesian_restaurant', 'malaysian_restaurant', 'filipino_restaurant', 'cambodian_restaurant', 'burmese_restaurant', 'tibetan_restaurant', 'taiwanese_restaurant', 'afghani_restaurant', 'african_restaurant', 'ethiopian_restaurant', 'moroccan_restaurant', 'lebanese_restaurant', 'israeli_restaurant', 'persian_restaurant', 'turkish_restaurant', 'middle_eastern_restaurant', 'mediterranean_restaurant', 'mexican_restaurant', 'latin_american_restaurant', 'south_american_restaurant', 'brazilian_restaurant', 'argentinian_restaurant', 'colombian_restaurant', 'chilean_restaurant', 'peruvian_restaurant', 'cuban_restaurant', 'caribbean_restaurant', 'hawaiian_restaurant', 'asian_restaurant', 'asian_fusion_restaurant', 'mongolian_barbecue_restaurant', 'pakistani_restaurant', 'bangladeshi_restaurant', 'sri_lankan_restaurant', 'fusion_restaurant', 'bistro', 'bakery', 'cake_shop', 'pastry_shop', 'dessert_restaurant', 'dessert_shop', 'donut_shop', 'ice_cream_shop', 'confectionery', 'chocolate_shop', 'bar', 'pub', 'night_club', 'cocktail_bar', 'wine_bar', 'lounge_bar', 'sports_bar', 'hookah_bar', 'irish_pub', 'beer_garden', 'karaoke', 'fast_food_restaurant', 'meal_takeaway', 'sandwich_shop', 'deli', 'snack_bar', 'burrito_restaurant', 'taco_restaurant', 'chicken_restaurant', 'kebab_shop', 'gyro_restaurant', 'shawarma_restaurant', 'hamburger_restaurant', 'vegan_restaurant', 'vegetarian_restaurant', 'seafood_restaurant', 'oyster_bar_restaurant', 'coffee_shop', 'cafe', 'coffee_roastery', 'coffee_stand', 'tea_house', 'juice_shop', 'hot_pot_restaurant', 'barbecue_restaurant', 'bar_and_grill', 'korean_barbecue_restaurant', 'noodle_shop', 'ramen_restaurant', 'chinese_noodle_restaurant', 'fine_dining_restaurant', 'buffet_restaurant', 'brunch_restaurant', 'breakfast_restaurant', 'halal_restaurant', 'gastropub', 'brewpub', 'brewery', 'winery', 'restaurant', 'food_court', 'cafeteria', 'diner')",
    'types[] có enum thuộc business_category = Bar & Pub/Nightlife servesBeer = true': "json_type in ('bar', 'pub', 'night_club', 'cocktail_bar', 'wine_bar', 'lounge_bar', 'sports_bar', 'hookah_bar', 'irish_pub', 'beer_garden', 'karaoke')",
    'types[] có enum thuộc business_category = Day Spa/Therapeutic Massage/Clinic-Aesthetic Spa/Hair Salon/Nail-Lash Salon/Beauty Salon/Barbershop': "json_type in ('spa', 'day_spa', 'wellness_center', 'massage', 'skin_care_clinic', 'acupuncture_clinic', 'ayurvedic_clinic', 'hair_salon', 'hair_care', 'nail_salon', 'beauty_salon', 'barber_shop')",
    'types[] có enum thuộc business_category = Gym-Fitness/Yoga-Pilates Studio': "json_type in ('gym', 'fitness_center', 'sports_club', 'recreation_center', 'yoga_studio', 'pilates_studio')",
    'types[] có electric_vehicle_charging_station': "json_type = 'electric_vehicle_charging_station'",
    'types[] có enum thuộc business_category = Historical Site': "json_type in ('historical_landmark', 'archaeological_site', 'ruins', 'castle', 'palace', 'monument', 'sculpture', 'lighthouse')",
    'types[] có art_gallery': "json_type = 'art_gallery'",
    'types[] có planetarium': "json_type = 'planetarium'",
    'types[] có business_category = Religious Site': "json_type in ('church', 'hindu_temple', 'mosque', 'synagogue', 'buddhist_temple', 'place_of_worship')",
    'types[] có restaurant, cafe, coffee_shop': "json_type in ('restaurant', 'cafe', 'coffee_shop')",
    'types[] có gift_shop': "json_type = 'gift_shop'",
    'types[] thuộc business_category = Bar & Pub/Nightlife': "json_type in ('bar', 'pub', 'night_club', 'cocktail_bar', 'wine_bar', 'lounge_bar', 'sports_bar', 'hookah_bar', 'irish_pub', 'beer_garden', 'karaoke')",
}


AMENITY_SHARED_RAW = (("google", "attraction", "facilities_and_services", "goodForChildren"),)


def amenity_keys(raw):
    """amenity_raw mapping -> list key. "a.b = true" -> ["a.b"]; "types[] = x, y" -> ["x", "y"]; câu types[] có/thuộc -> json_type."""
    mapped = AMENITY_JSON_TYPE.get(raw)
    if mapped:
        return [mapped]
    flag = AMENITY_FLAG_RE.fullmatch(raw)
    if flag:
        return [flag.group(1)]
    types = AMENITY_TYPES_RE.fullmatch(raw)
    if types:
        return [part.strip() for part in types.group(1).split(",")]
    return [raw]

SEED_SECTORS = ("accommodation", "f_n_b", "attraction", "shopping_n_retail")
SEED_GROUPS = (
    "general_amenities",
    "accessibility",
    "specialty_tags",
    "space_and_services",
    "facilities_and_services",
)
SEED_SKIP = ("Bổ sung sau", "Fallback", "Không có field", "Không có field riêng")
SEED_ROWS = [
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.freeStreetParking', 'Free Street Parking', 'Đỗ xe trên phố'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.valetParking', 'Valet Parking', 'Dịch vụ Valet'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.freeGarageParking', 'Free Garage Parking', 'Bãi xe trong nhà (Miễn phí)'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.paidGarageParking', 'Paid Garage Parking', 'Bãi xe trong nhà (Thu phí)'),
 ('google', 'accommodation', 'general_amenities', 'outdoorSeating', 'Outdoor Seating', 'Không gian ngoài trời'),
 ('google', 'accommodation', 'general_amenities', 'restroom', 'Restroom', 'Nhà vệ sinh'),
 ('google', 'accommodation', 'general_amenities', 'goodForChildren', 'Good for Children', 'Thân thiện trẻ em'),
 ('google', 'accommodation', 'general_amenities', 'goodForGroups', 'Good for Groups', 'Phù hợp đoàn nhóm'),
 ('google', 'accommodation', 'general_amenities', 'liveMusic', 'Live Music', 'Nhạc Live'),
 ('partner', 'accommodation', 'general_amenities', 'Hồ bơi', 'Swimming Pool', 'Hồ bơi'),
 ('google', 'accommodation', 'general_amenities', 'swimming_pool', 'Swimming Pool', 'Hồ bơi'),
 ('partner', 'accommodation', 'general_amenities', 'Wi-Fi miễn phí', 'Free Wi-Fi', 'Wi-Fi miễn phí'),
 ('partner', 'accommodation', 'general_amenities', 'Bãi đậu xe', 'Parking', 'Bãi đậu xe'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.freeParkingLot', 'Parking', 'Bãi đậu xe'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.paidParkingLot', 'Parking', 'Bãi đậu xe'),
 ('partner', 'accommodation', 'general_amenities', 'Xe đưa đón sân bay', 'Airport Shuttle', 'Xe đưa đón sân bay'),
 ('partner', 'accommodation', 'general_amenities', 'Nhà hàng', 'Restaurant', 'Nhà hàng'),
 ('google', 'accommodation', 'general_amenities', "json_type in ('vietnamese_restaurant', 'chinese_restaurant', 'japanese_restaurant', 'korean_restaurant', 'thai_restaurant', 'french_restaurant', 'italian_restaurant', 'pizza_restaurant', 'american_restaurant', 'european_restaurant', 'british_restaurant', 'german_restaurant', 'spanish_restaurant', 'greek_restaurant', 'steak_house', 'portuguese_restaurant', 'austrian_restaurant', 'bavarian_restaurant', 'belgian_restaurant', 'danish_restaurant', 'dutch_restaurant', 'scandinavian_restaurant', 'swiss_restaurant', 'polish_restaurant', 'romanian_restaurant', 'russian_restaurant', 'ukrainian_restaurant', 'croatian_restaurant', 'czech_restaurant', 'hungarian_restaurant', 'australian_restaurant', 'californian_restaurant', 'tex_mex_restaurant', 'fish_and_chips_restaurant', 'basque_restaurant', 'irish_restaurant', 'soul_food_restaurant', 'southwestern_us_restaurant', 'eastern_european_restaurant', 'indian_restaurant', 'north_indian_restaurant', 'south_indian_restaurant', 'indonesian_restaurant', 'malaysian_restaurant', 'filipino_restaurant', 'cambodian_restaurant', 'burmese_restaurant', 'tibetan_restaurant', 'taiwanese_restaurant', 'afghani_restaurant', 'african_restaurant', 'ethiopian_restaurant', 'moroccan_restaurant', 'lebanese_restaurant', 'israeli_restaurant', 'persian_restaurant', 'turkish_restaurant', 'middle_eastern_restaurant', 'mediterranean_restaurant', 'mexican_restaurant', 'latin_american_restaurant', 'south_american_restaurant', 'brazilian_restaurant', 'argentinian_restaurant', 'colombian_restaurant', 'chilean_restaurant', 'peruvian_restaurant', 'cuban_restaurant', 'caribbean_restaurant', 'hawaiian_restaurant', 'asian_restaurant', 'asian_fusion_restaurant', 'mongolian_barbecue_restaurant', 'pakistani_restaurant', 'bangladeshi_restaurant', 'sri_lankan_restaurant', 'fusion_restaurant', 'bistro', 'bakery', 'cake_shop', 'pastry_shop', 'dessert_restaurant', 'dessert_shop', 'donut_shop', 'ice_cream_shop', 'confectionery', 'chocolate_shop', 'bar', 'pub', 'night_club', 'cocktail_bar', 'wine_bar', 'lounge_bar', 'sports_bar', 'hookah_bar', 'irish_pub', 'beer_garden', 'karaoke', 'fast_food_restaurant', 'meal_takeaway', 'sandwich_shop', 'deli', 'snack_bar', 'burrito_restaurant', 'taco_restaurant', 'chicken_restaurant', 'kebab_shop', 'gyro_restaurant', 'shawarma_restaurant', 'hamburger_restaurant', 'vegan_restaurant', 'vegetarian_restaurant', 'seafood_restaurant', 'oyster_bar_restaurant', 'coffee_shop', 'cafe', 'coffee_roastery', 'coffee_stand', 'tea_house', 'juice_shop', 'hot_pot_restaurant', 'barbecue_restaurant', 'bar_and_grill', 'korean_barbecue_restaurant', 'noodle_shop', 'ramen_restaurant', 'chinese_noodle_restaurant', 'fine_dining_restaurant', 'buffet_restaurant', 'brunch_restaurant', 'breakfast_restaurant', 'halal_restaurant', 'gastropub', 'brewpub', 'brewery', 'winery', 'restaurant', 'food_court', 'cafeteria', 'diner')", 'Restaurant', 'Nhà hàng'),
 ('google', 'accommodation', 'general_amenities', 'serves_meals = true', 'Restaurant', 'Nhà hàng'),
 ('partner', 'accommodation', 'general_amenities', 'Lễ tân 24h', '24-hour Front Desk', 'Lễ tân 24h'),
 ('partner', 'accommodation', 'general_amenities', 'Quầy bar', 'Bar', 'Quầy bar'),
 ('google', 'accommodation', 'general_amenities', "json_type in ('bar', 'pub', 'night_club', 'cocktail_bar', 'wine_bar', 'lounge_bar', 'sports_bar', 'hookah_bar', 'irish_pub', 'beer_garden', 'karaoke')", 'Bar', 'Quầy bar'),
 ('google', 'accommodation', 'general_amenities', 'servesCocktails', 'Bar', 'Quầy bar'),
 ('partner', 'accommodation', 'general_amenities', 'Spa & Wellness', 'Spa & Wellness', 'Spa & Wellness'),
 ('google', 'accommodation', 'general_amenities', "json_type in ('spa', 'day_spa', 'wellness_center', 'massage', 'skin_care_clinic', 'acupuncture_clinic', 'ayurvedic_clinic', 'hair_salon', 'hair_care', 'nail_salon', 'beauty_salon', 'barber_shop')", 'Spa & Wellness', 'Spa & Wellness'),
 ('partner', 'accommodation', 'general_amenities', 'Phòng Gym', 'Gym', 'Phòng Gym'),
 ('google', 'accommodation', 'general_amenities', "json_type in ('gym', 'fitness_center', 'sports_club', 'recreation_center', 'yoga_studio', 'pilates_studio')", 'Gym', 'Phòng Gym'),
 ('partner', 'accommodation', 'general_amenities', 'Xông hơi', 'Sauna', 'Xông hơi'),
 ('google', 'accommodation', 'general_amenities', 'hot_spring', 'Sauna', 'Xông hơi'),
 ('google', 'accommodation', 'general_amenities', 'sauna', 'Sauna', 'Xông hơi'),
 ('partner', 'accommodation', 'general_amenities', 'Bồn tắm nóng (Jacuzzi)', 'Jacuzzi', 'Bồn tắm nóng'),
 ('partner', 'accommodation', 'general_amenities', 'Sân vườn', 'Garden', 'Sân vườn'),
 ('google', 'accommodation', 'general_amenities', 'garden', 'Garden', 'Sân vườn'),
 ('google', 'accommodation', 'general_amenities', 'botanical_garden', 'Garden', 'Sân vườn'),
 ('partner', 'accommodation', 'general_amenities', 'Phòng họp', 'Meeting Rooms', 'Phòng họp'),
 ('google', 'accommodation', 'general_amenities', 'meeting_rooms', 'Meeting Rooms', 'Phòng họp'),
 ('partner', 'accommodation', 'general_amenities', 'Khu vực làm việc chung', 'Co-working Space', 'Khu làm việc chung'),
 ('partner', 'accommodation', 'general_amenities', 'Giữ hành lý', 'Luggage Storage', 'Giữ hành lý'),
 ('partner', 'accommodation', 'general_amenities', 'Thu đổi ngoại tệ', 'Currency Exchange', 'Thu đổi ngoại tệ'),
 ('partner', 'accommodation', 'general_amenities', 'Trạm sạc xe điện', 'EV Charging Station', 'Trạm sạc xe điện'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.freeElectricVehicleParking', 'EV Charging Station', 'Trạm sạc xe điện'),
 ('google', 'accommodation', 'general_amenities', 'parkingOptions.paidElectricVehicleParking', 'EV Charging Station', 'Trạm sạc xe điện'),
 ('google', 'accommodation', 'general_amenities', "json_type = 'electric_vehicle_charging_station'", 'EV Charging Station', 'Trạm sạc xe điện'),
 ('google', 'accommodation', 'accessibility', 'accessibilityOptions.wheelchairAccessibleSeating', 'Wheelchair Accessible Seating', 'Chỗ ngồi cho người khuyết tật'),
 ('google', 'accommodation', 'accessibility', 'accessibilityOptions.wheelchairAccessibleElevator', 'Wheelchair Accessible Elevator', 'Thang máy hỗ trợ xe lăn'),
 ('partner', 'accommodation', 'accessibility', 'Lối đi xe lăn', 'Wheelchair Accessible Entrance', 'Lối đi xe lăn'),
 ('google', 'accommodation', 'accessibility', 'accessibilityOptions.wheelchairAccessibleEntrance', 'Wheelchair Accessible Entrance', 'Lối đi xe lăn'),
 ('partner', 'accommodation', 'accessibility', 'Thang máy có chữ nổi', 'Braille Elevator', 'Thang máy có chữ nổi'),
 ('partner', 'accommodation', 'accessibility', 'Nhà vệ sinh có tay vịn', 'Wheelchair Accessible Restroom', 'Nhà vệ sinh có tay vịn'),
 ('google', 'accommodation', 'accessibility', 'accessibilityOptions.wheelchairAccessibleRestroom', 'Wheelchair Accessible Restroom', 'Nhà vệ sinh có tay vịn'),
 ('partner', 'accommodation', 'accessibility', 'Bồn rửa mặt thấp', 'Lowered Sink', 'Bồn rửa mặt thấp'),
 ('partner', 'accommodation', 'accessibility', 'Bãi đỗ xe cho người khuyết tật', 'Accessible Parking', 'Bãi đỗ xe cho người khuyết tật'),
 ('google', 'accommodation', 'accessibility', 'accessibilityOptions.wheelchairAccessibleParking', 'Accessible Parking', 'Bãi đỗ xe cho người khuyết tật'),
 ('google', 'f_n_b', 'specialty_tags', 'dim_sum_restaurant', 'Dim Sum', 'Dimsum'),
 ('google', 'f_n_b', 'specialty_tags', 'sushi_restaurant', 'Sushi', 'Sushi'),
 ('google', 'f_n_b', 'specialty_tags', 'dumpling_restaurant', 'Dumpling', 'Há cảo – Bánh bao'),
 ('google', 'f_n_b', 'specialty_tags', 'tonkatsu_restaurant', 'Tonkatsu', 'Tonkatsu'),
 ('google', 'f_n_b', 'specialty_tags', 'japanese_curry_restaurant', 'Japanese Curry', 'Cà ri Nhật'),
 ('google', 'f_n_b', 'specialty_tags', 'cantonese_restaurant', 'Cantonese', 'Ẩm thực Quảng Đông'),
 ('google', 'f_n_b', 'specialty_tags', 'chinese_noodle_restaurant', 'Chinese Noodle', 'Mì Trung Quốc'),
 ('google', 'f_n_b', 'specialty_tags', 'ramen_restaurant', 'Ramen', 'Ramen'),
 ('google', 'f_n_b', 'specialty_tags', 'yakiniku_restaurant, yakitori_restaurant', 'Yakiniku / Teppanyaki', 'Yakiniku / Teppanyaki'),
 ('google', 'f_n_b', 'specialty_tags', 'japanese_izakaya_restaurant', 'Izakaya', 'Izakaya'),
 ('google', 'f_n_b', 'specialty_tags', 'korean_barbecue_restaurant', 'Korean BBQ', 'BBQ Hàn Quốc'),
 ('google', 'f_n_b', 'specialty_tags', 'pizza_restaurant', 'Pizza', 'Pizza'),
 ('google', 'f_n_b', 'specialty_tags', 'hamburger_restaurant', 'Hamburger', 'Hamburger'),
 ('google', 'f_n_b', 'specialty_tags', 'brunch_restaurant', 'Brunch', 'Brunch'),
 ('google', 'f_n_b', 'specialty_tags', 'cat_cafe, dog_cafe', 'Pet Café', 'Quán cà phê thú cưng'),
 ('google', 'f_n_b', 'specialty_tags', 'tapas_restaurant', 'Tapas', 'Tapas'),
 ('google', 'f_n_b', 'specialty_tags', 'soup_restaurant', 'Soup', 'Súp'),
 ('partner', 'f_n_b', 'specialty_tags', 'Street Food', 'Street Food', 'Ẩm thực đường phố'),
 ('google', 'f_n_b', 'specialty_tags', 'food_stand', 'Street Food', 'Ẩm thực đường phố'),
 ('google', 'f_n_b', 'specialty_tags', 'street_food_gathering', 'Street Food', 'Ẩm thực đường phố'),
 ('partner', 'f_n_b', 'specialty_tags', 'Seafood', 'Seafood', 'Hải sản'),
 ('google', 'f_n_b', 'specialty_tags', 'seafood_restaurant', 'Seafood', 'Hải sản'),
 ('google', 'f_n_b', 'specialty_tags', 'oyster_bar_restaurant', 'Seafood', 'Hải sản'),
 ('partner', 'f_n_b', 'specialty_tags', 'Steakhouse', 'Steakhouse', 'Bít tết'),
 ('google', 'f_n_b', 'specialty_tags', 'steakhouse', 'Steakhouse', 'Bít tết'),
 ('partner', 'f_n_b', 'specialty_tags', 'Office Lunch', 'Office Lunch', 'Cơm văn phòng'),
 ('partner', 'f_n_b', 'specialty_tags', 'Milk Tea', 'Milk Tea', 'Trà sữa'),
 ('partner', 'f_n_b', 'specialty_tags', 'Bakery', 'Bakery', 'Tiệm bánh'),
 ('google', 'f_n_b', 'specialty_tags', 'bakery', 'Bakery', 'Tiệm bánh'),
 ('google', 'f_n_b', 'specialty_tags', 'pastry_shop', 'Bakery', 'Tiệm bánh'),
 ('google', 'f_n_b', 'specialty_tags', 'cake_shop', 'Bakery', 'Tiệm bánh'),
 ('partner', 'f_n_b', 'specialty_tags', 'Bar-Pub', 'Bar & Pub', 'Bar – Pub'),
 ('google', 'f_n_b', 'specialty_tags', "json_type in ('bar', 'pub', 'night_club', 'cocktail_bar', 'wine_bar', 'lounge_bar', 'sports_bar', 'hookah_bar', 'irish_pub', 'beer_garden', 'karaoke')", 'Bar & Pub', 'Bar – Pub'),
 ('partner', 'f_n_b', 'specialty_tags', 'Rooftop Bar', 'Rooftop Bar', 'Bar sân thượng'),
 ('google', 'f_n_b', 'space_and_services', 'restroom', 'Restroom', 'Nhà vệ sinh'),
 ('google', 'f_n_b', 'space_and_services', 'goodForChildren', 'Good for Children', 'Thân thiện trẻ em'),
 ('partner', 'f_n_b', 'space_and_services', 'Outdoor Seating', 'Outdoor Seating', 'Chỗ ngồi ngoài trời'),
 ('google', 'f_n_b', 'space_and_services', 'outdoorSeating = true = true', 'Outdoor Seating', 'Chỗ ngồi ngoài trời'),
 ('partner', 'f_n_b', 'space_and_services', 'Private Room (VIP)', 'Private Room (VIP)', 'Phòng riêng (VIP)'),
 ('partner', 'f_n_b', 'space_and_services', 'Scenic View', 'Scenic View', 'Góc nhìn đẹp'),
 ('partner', 'f_n_b', 'space_and_services', 'Romantic Decor', 'Romantic Decor', 'Trang trí lãng mạn'),
 ('partner', 'f_n_b', 'space_and_services', 'High Chair', 'High Chair', 'Có ghế trẻ em'),
 ('partner', 'f_n_b', 'space_and_services', 'Wheelchair Accessible', 'Wheelchair Accessible', 'Lối đi xe lăn'),
 ('google', 'f_n_b', 'space_and_services', 'accessibilityOptions.wheelchairAccessibleEntrance', 'Wheelchair Accessible', 'Lối đi xe lăn'),
 ('partner', 'f_n_b', 'space_and_services', 'Sports Screening', 'Sports Screening', 'Chiếu thể thao trực tiếp'),
 ('google', 'f_n_b', 'space_and_services', 'goodForWatchingSports', 'Sports Screening', 'Chiếu thể thao trực tiếp'),
 ('partner', 'f_n_b', 'space_and_services', 'Live Music', 'Live Music', 'Nhạc sống'),
 ('google', 'f_n_b', 'space_and_services', 'liveMusic', 'Live Music', 'Nhạc sống'),
 ('partner', 'f_n_b', 'space_and_services', 'Car Parking', 'Car Parking', 'Bãi đỗ xe ô tô'),
 ('google', 'f_n_b', 'space_and_services', 'parkingOptions.freeParkingLot, parkingOptions.paidParkingLot = true', 'Car Parking', 'Bãi đỗ xe ô tô'),
 ('partner', 'f_n_b', 'space_and_services', 'Table Service', 'Table Service', 'Phục vụ tại bàn'),
 ('google', 'f_n_b', 'space_and_services', 'dineIn', 'Table Service', 'Phục vụ tại bàn'),
 ('google', 'attraction', 'facilities_and_services', 'accessibilityOptions.wheelchairAccessibleEntrance', 'Wheelchair Accessible Entrance', 'Lối đi xe lăn'),
 ('google', 'attraction', 'facilities_and_services', 'accessibilityOptions.wheelchairAccessibleParking', 'Accessible Parking', 'Bãi đỗ xe người khuyết tật'),
 ('google', 'attraction', 'facilities_and_services', 'accessibilityOptions.wheelchairAccessibleRestroom', 'Wheelchair Accessible Restroom', 'Nhà vệ sinh xe lăn'),
 ('google', 'attraction', 'facilities_and_services', 'restroom', 'Restroom', 'Nhà vệ sinh'),
 ('google', 'attraction', 'facilities_and_services', 'goodForChildren', 'Good for Children', 'Thân thiện trẻ em'),
 ('google', 'attraction', 'facilities_and_services', 'liveMusic', 'Live Music', 'Nhạc Live'),
 ('partner', 'attraction', 'facilities_and_services', 'Indoor', 'Indoor', 'Trong nhà'),
 ('partner', 'attraction', 'facilities_and_services', 'Outdoor', 'Outdoor', 'Ngoài trời'),
 ('partner', 'attraction', 'facilities_and_services', 'Covered', 'Covered', 'Có mái che'),
 ('partner', 'attraction', 'facilities_and_services', 'Family Friendly', 'Family Friendly', 'Thân thiện với gia đình'),
 ('google', 'attraction', 'facilities_and_services', 'goodForChildren', 'Family Friendly', 'Thân thiện với gia đình'),
 ('partner', 'attraction', 'facilities_and_services', 'Couples', 'Couples', 'Cặp đôi'),
 ('partner', 'attraction', 'facilities_and_services', 'Group', 'Group', 'Nhóm bạn'),
 ('google', 'attraction', 'facilities_and_services', 'goodForGroups', 'Group', 'Nhóm bạn'),
 ('partner', 'attraction', 'facilities_and_services', 'History', 'History', 'Lịch sử'),
 ('google', 'attraction', 'facilities_and_services', "json_type in ('historical_landmark', 'archaeological_site', 'ruins', 'castle', 'palace', 'monument', 'sculpture', 'lighthouse')", 'History', 'Lịch sử'),
 ('partner', 'attraction', 'facilities_and_services', 'Art', 'Art', 'Nghệ thuật'),
 ('google', 'attraction', 'facilities_and_services', "json_type = 'art_gallery'", 'Art', 'Nghệ thuật'),
 ('partner', 'attraction', 'facilities_and_services', 'Science', 'Science', 'Khoa học'),
 ('google', 'attraction', 'facilities_and_services', "json_type = 'planetarium'", 'Science', 'Khoa học'),
 ('partner', 'attraction', 'facilities_and_services', 'Interactive', 'Interactive', 'Tương tác'),
 ('partner', 'attraction', 'facilities_and_services', 'Educational', 'Educational', 'Giáo dục'),
 ('partner', 'attraction', 'facilities_and_services', 'Spiritual', 'Spiritual', 'Tâm linh'),
 ('google', 'attraction', 'facilities_and_services', "json_type in ('church', 'hindu_temple', 'mosque', 'synagogue', 'buddhist_temple', 'place_of_worship')", 'Spiritual', 'Tâm linh'),
 ('partner', 'attraction', 'facilities_and_services', 'Parking', 'Parking', 'Bãi đậu xe'),
 ('google', 'attraction', 'facilities_and_services', 'parkingOptions.freeParkingLot', 'Parking', 'Bãi đậu xe'),
 ('google', 'attraction', 'facilities_and_services', 'parkingOptions.paidParkingLot', 'Parking', 'Bãi đậu xe'),
 ('partner', 'attraction', 'facilities_and_services', 'Free Wi-Fi', 'Free Wi-Fi', 'Wi-Fi miễn phí'),
 ('partner', 'attraction', 'facilities_and_services', 'Lockers', 'Lockers', 'Tủ gửi đồ'),
 ('partner', 'attraction', 'facilities_and_services', 'Restaurant-Cafe', 'Restaurant-Cafe', 'Nhà hàng – Quán ăn'),
 ('google', 'attraction', 'facilities_and_services', "json_type in ('restaurant', 'cafe', 'coffee_shop')", 'Restaurant-Cafe', 'Nhà hàng – Quán ăn'),
 ('partner', 'attraction', 'facilities_and_services', 'Gift Shop', 'Gift Shop', 'Cửa hàng lưu niệm'),
 ('google', 'attraction', 'facilities_and_services', "json_type = 'gift_shop'", 'Gift Shop', 'Cửa hàng lưu niệm'),
 ('partner', 'attraction', 'facilities_and_services', 'Stroller Rental', 'Stroller Rental', 'Cho thuê xe đẩy trẻ em'),
 ('partner', 'attraction', 'facilities_and_services', 'Wheelchair Rental', 'Wheelchair Rental', 'Cho thuê xe lăn'),
 ('partner', 'attraction', 'facilities_and_services', 'First Aid Room', 'First Aid Room', 'Phòng y tế'),
 ('partner', 'attraction', 'facilities_and_services', 'On-site Tour Guide', 'On-site Tour Guide', 'Hướng dẫn viên tại điểm'),
 ('partner', 'attraction', 'facilities_and_services', 'Audio Guide', 'Audio Guide', 'Máy hướng dẫn tự động'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Tax Refund', 'Tax Refund', 'Hoàn thuế'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Shipping-Delivery', 'Shipping-Delivery', 'Giao hàng tận nơi'),
 ('google', 'shopping_n_retail', 'facilities_and_services', 'delivery', 'Shipping-Delivery', 'Giao hàng tận nơi'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Gift Wrapping', 'Gift Wrapping', 'Gói quà miễn phí'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Fitting Room', 'Fitting Room', 'Phòng thử đồ'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Personalization-Engraving', 'Personalization-Engraving', 'Khắc tên – Cá nhân hóa'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Tailoring', 'Tailoring', 'May đo theo yêu cầu'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Credit Card (Visa/Master)', 'Credit Card', 'Chấp nhận thẻ tín dụng'),
 ('google', 'shopping_n_retail', 'facilities_and_services', 'paymentOptions.acceptsCreditCards', 'Credit Card', 'Chấp nhận thẻ tín dụng'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'E-Wallet', 'E-Wallet', 'Ví điện tử'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'VAT Invoice', 'VAT Invoice', 'Xuất hóa đơn VAT'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Flexible Return', 'Flexible Return', 'Đổi trả linh hoạt'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Parking', 'Parking', 'Bãi đỗ xe'),
 ('google', 'shopping_n_retail', 'facilities_and_services', 'parkingOptions.freeParkingLot, parkingOptions.paidParkingLot', 'Parking', 'Bãi đỗ xe'),
 ('partner', 'shopping_n_retail', 'facilities_and_services', 'Free Wi-Fi', 'Free Wi-Fi', 'Wi-Fi miễn phí'),
]

def validate_seed(rows):
    """Seed phải đúng nguồn, sector, nhóm, khoá 5 cột, chỉ còn dạng key; khoá 4 cột trùng chỉ trong AMENITY_SHARED_RAW."""
    keys = [row[:5] for row in rows]
    if len(keys) != len(set(keys)):
        raise ValueError("ref_dictionary_amenities: trùng khoá 5 cột")
    counts = {}
    for row in rows:
        counts[row[:4]] = counts.get(row[:4], 0) + 1
    dup = sorted(key for key, n in counts.items() if n > 1 and key not in AMENITY_SHARED_RAW)
    if dup:
        raise ValueError(f"ref_dictionary_amenities: trùng amenity_raw trong (source, sector, group): {dup}")
    for source, sector, group, raw, amenity, _vi in rows:
        if source not in {"google", "partner"}:
            raise ValueError(f"source lạ: {source}")
        if sector not in SEED_SECTORS:
            raise ValueError(f"sector lạ: {sector}")
        if group not in SEED_GROUPS:
            raise ValueError(f"group lạ: {group}")
        if not raw or raw in SEED_SKIP:
            raise ValueError(f"amenity_raw không hợp lệ: {raw}")
        if amenity_keys(raw) != [raw]:
            raise ValueError(f"amenity_raw còn dạng cũ: {raw}")
        if not amenity:
            raise ValueError(f"amenity trống: {raw}")
# END_AMENITY_SEED
```

```python
def _seed_merge():
    spark.sql("""
        MERGE INTO lh_vv_silver.dbo.ref_dictionary_amenities AS t
        USING seed_amenities AS s
        ON t.source_name = s.source_name
       AND t.business_sector = s.business_sector
       AND t.amenity_group = s.amenity_group
       AND t.amenity_raw = s.amenity_raw
       AND t.amenity = s.amenity
        WHEN NOT MATCHED THEN INSERT (
            source_name, business_sector, amenity_group, amenity_raw, amenity, amenity_vi, is_active, updated_at
        ) VALUES (
            s.source_name, s.business_sector, s.amenity_group, s.amenity_raw, s.amenity, s.amenity_vi,
            true, current_timestamp()
        )
    """)


missing_before = spark.sql("""
    SELECT count(*) AS n
    FROM seed_amenities s
    LEFT ANTI JOIN lh_vv_silver.dbo.ref_dictionary_amenities t
      ON t.source_name = s.source_name
     AND t.business_sector = s.business_sector
     AND t.amenity_group = s.amenity_group
     AND t.amenity_raw = s.amenity_raw
     AND t.amenity = s.amenity
""").first()["n"]
with_retry(_seed_merge, "seed ref_dictionary_amenities")
INSERTED = int(missing_before)

dups = spark.sql("""
    SELECT count(*) AS n FROM (
        SELECT source_name, business_sector, amenity_group, amenity_raw, amenity
        FROM lh_vv_silver.dbo.ref_dictionary_amenities
        GROUP BY source_name, business_sector, amenity_group, amenity_raw, amenity
        HAVING count(*) > 1
    )
""").first()["n"]
if dups:
    raise ConfigError(f"ref_dictionary_amenities: {dups} khoá 5 cột trùng")

missing_after = spark.sql("""
    SELECT count(*) AS n
    FROM seed_amenities s
    LEFT ANTI JOIN lh_vv_silver.dbo.ref_dictionary_amenities t
      ON t.source_name = s.source_name
     AND t.business_sector = s.business_sector
     AND t.amenity_group = s.amenity_group
     AND t.amenity_raw = s.amenity_raw
     AND t.amenity = s.amenity
""").first()["n"]
if missing_after:
    raise ConfigError(f"ref_dictionary_amenities: thiếu {missing_after} khoá seed")

DRIFT = int(spark.sql("""
    SELECT count(*) AS n
    FROM seed_amenities s
    JOIN lh_vv_silver.dbo.ref_dictionary_amenities t
      ON t.source_name = s.source_name
     AND t.business_sector = s.business_sector
     AND t.amenity_group = s.amenity_group
     AND t.amenity_raw = s.amenity_raw
     AND t.amenity = s.amenity
    WHERE NOT (t.amenity_vi <=> s.amenity_vi AND t.is_active = true)
""").first()["n"])
STALE = int(spark.sql(STALE_SQL).first()["n"])
print(f"inserted={INSERTED} drift={DRIFT} stale={STALE}")
```

```python
import uuid

# BEGIN_EDGE_PRECHECK
def setup_edge_issues(rows, upserts, retires):
    """Cạnh active trên lh_vv_silver.dbo mà MERGE không ghi thì chặn. Đúng bộ RETIRE thì cho qua.

    rows: pl_name, src_schema, src_tbl, trg_schema, trg_tbl, is_active.
    upserts, retires: tập (src_tbl, trg_tbl) của MERGE này, schema lh_vv_silver.dbo, pl SLV_1H.
    Schema khác không tính. Output: list thông báo, rỗng nghĩa là được MERGE.
    """
    silver = "lh_vv_silver.dbo"
    flow = "PL_VV_TRANS_SLV_TO_SLV_1H"
    targets = {trg for _src, trg in upserts}
    extra, stale_entity, stale_amenity = [], 0, 0
    for row in rows:
        if int(row.get("is_active") or 0) != 1:
            continue
        if row.get("src_schema") != silver or row.get("trg_schema") != silver:
            continue
        key = (row["src_tbl"], row["trg_tbl"])
        planned_retire = row.get("pl_name") == flow and key in retires
        if row["trg_tbl"] in targets and key not in upserts and not planned_retire:
            extra.append(f"{row['trg_tbl']}/{row['pl_name']}/{row['src_tbl']}")
        if row["trg_tbl"] == "slv_poi_entity" and not planned_retire:
            stale_entity += 1
        if key == ("slv_poi_entity", "slv_poi_amenity") and row.get("pl_name") != flow:
            stale_amenity += 1
    issues = []
    if extra:
        issues.append("còn cạnh active ngoài cạnh chuẩn: " + ", ".join(extra))
    if stale_entity:
        issues.append(f"slv_poi_entity: còn {stale_entity} cạnh active")
    if stale_amenity:
        issues.append("slv_poi_amenity: còn cạnh active slv_poi_entity")
    return issues
# END_EDGE_PRECHECK


def _write_edges():
    """Ghi cạnh amenity trong một MERGE. Kiểm tra dưới lock, trước khi commit."""
    cfg = "lh_vv_ctrl.dbo.ctrl_mng_pipeline_config"
    has_nb = "nb_name" in table_columns(cfg)
    nb_match = " AND t.nb_name <=> CAST(NULL AS STRING)" if has_nb else ""
    nb_set = ", nb_name = NULL" if has_nb else ""
    nb_col = ", nb_name" if has_nb else ""
    nb_val = ", CAST(NULL AS STRING)" if has_nb else ""
    entity_edges_ok = spark.sql(f"""
        SELECT count(*) AS n
        FROM {cfg}
        WHERE pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
          AND src_schema = 'lh_vv_silver.dbo'
          AND trg_schema = 'lh_vv_silver.dbo'
          AND trg_tbl = 'slv_poi_entity_non_enrich'
          AND is_active = 1
          AND src_tbl IN (
            'slv_3p_poi', 'slv_3p_poi_address', 'ref_dictionary_place_type',
            'slv_pn_business_services', 'slv_pn_business_service_i18ns')
    """).first()["n"]
    if entity_edges_ok != 5:
        raise ConfigError("slv_poi_amenity: chưa có cạnh slv_poi_entity_non_enrich, chạy NB_SETUP_POI_ENTITY trước")
    edge_rows = """
            (1, 'lh_vv_silver.dbo', 'slv_3p_poi_amenity', 'lh_vv_silver.dbo', 'slv_poi_amenity', 'UPSERT'),
            (2, 'lh_vv_silver.dbo', 'slv_poi_entity_non_enrich', 'lh_vv_silver.dbo', 'slv_poi_amenity', 'UPSERT'),
            (3, 'lh_vv_silver.dbo', 'ref_dictionary_amenities', 'lh_vv_silver.dbo', 'slv_poi_amenity', 'UPSERT'),
            (0, 'lh_vv_silver.dbo', 'slv_poi_entity', 'lh_vv_silver.dbo', 'slv_poi_amenity', 'RETIRE')"""
    dup = spark.sql(f"""
        SELECT trg_tbl, src_tbl, count(*) AS n
        FROM {cfg}
        WHERE pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
          AND src_schema = 'lh_vv_silver.dbo'
          AND trg_schema = 'lh_vv_silver.dbo'
          AND trg_tbl = 'slv_poi_amenity'
          AND src_tbl IN (
            'slv_3p_poi_amenity', 'slv_poi_entity_non_enrich', 'ref_dictionary_amenities', 'slv_poi_entity')
        GROUP BY src_schema, src_tbl, trg_schema, trg_tbl
        HAVING count(*) > 1
    """).collect()
    if dup:
        names = ", ".join(f"{row.trg_tbl}/{row.src_tbl}={row.n}" for row in dup)
        raise ConfigError(f"slv_poi_amenity: trùng cạnh, không ghi đè: {names}")
    owners = spark.sql(f"""
        SELECT DISTINCT pl_name
        FROM {cfg}
        WHERE trg_schema = 'lh_vv_silver.dbo'
          AND trg_tbl = 'slv_poi_amenity'
          AND is_active = 1
          AND pl_name <> 'PL_VV_TRANS_SLV_TO_SLV_1H'
    """).collect()
    if owners:
        names = ", ".join(row.pl_name for row in owners)
        raise ConfigError(f"slv_poi_amenity đã thuộc pl_name khác: {names}")
    watched = spark.sql(f"""
        SELECT pl_name, src_schema, src_tbl, trg_schema, trg_tbl, is_active
        FROM {cfg}
        WHERE src_schema = 'lh_vv_silver.dbo' AND trg_schema = 'lh_vv_silver.dbo'
          AND (
            trg_tbl IN ('slv_poi_amenity', 'slv_poi_entity')
            OR src_tbl = 'slv_poi_entity'
          )
    """).collect()
    amenity_upserts = {
        ("slv_3p_poi_amenity", "slv_poi_amenity"),
        ("slv_poi_entity_non_enrich", "slv_poi_amenity"),
        ("ref_dictionary_amenities", "slv_poi_amenity"),
    }
    amenity_retires = {("slv_poi_entity", "slv_poi_amenity")}
    issues = setup_edge_issues([row.asDict() for row in watched], amenity_upserts, amenity_retires)
    if issues:
        raise ConfigError("; ".join(issues))

    missing = spark.sql(f"""
        SELECT count(*) AS n
        FROM VALUES {edge_rows} AS t(seq, src_schema, src_tbl, trg_schema, trg_tbl, action)
        WHERE t.action = 'UPSERT' AND NOT EXISTS (
            SELECT 1 FROM {cfg} c
            WHERE c.pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
              AND c.src_schema = t.src_schema AND c.src_tbl = t.src_tbl
              AND c.trg_schema = t.trg_schema AND c.trg_tbl = t.trg_tbl)
    """).first()["n"]
    retired = spark.sql(f"""
        SELECT count(*) AS n FROM {cfg}
        WHERE pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
          AND src_schema = 'lh_vv_silver.dbo'
          AND trg_schema = 'lh_vv_silver.dbo'
          AND trg_tbl = 'slv_poi_amenity' AND src_tbl = 'slv_poi_entity'
          AND NOT (is_active <=> 0)
    """).first()["n"]
    spark.sql(f"""
        MERGE INTO {cfg} AS t
        USING (
            SELECT seq, src_schema, src_tbl, trg_schema, trg_tbl, action
            FROM VALUES {edge_rows} AS v(seq, src_schema, src_tbl, trg_schema, trg_tbl, action)
        ) AS s
        ON t.pl_name = 'PL_VV_TRANS_SLV_TO_SLV_1H'
           AND t.src_schema = s.src_schema
           AND t.src_tbl = s.src_tbl
           AND t.trg_schema = s.trg_schema
           AND t.trg_tbl = s.trg_tbl
        WHEN MATCHED AND s.action = 'RETIRE' AND NOT (t.is_active <=> 0) THEN UPDATE SET
            is_active = 0
        WHEN MATCHED AND s.action = 'UPSERT' AND NOT (
            t.is_active <=> 1 AND t.priority <=> 1 AND t.load_mode <=> 'RECOMPUTE'
            AND t.src_schema <=> 'lh_vv_silver.dbo' AND t.trg_schema <=> 'lh_vv_silver.dbo'
            AND t.align_path <=> CAST(NULL AS STRING) AND t.dedup_order <=> CAST(NULL AS STRING){nb_match}
        ) THEN UPDATE SET
            src_schema = s.src_schema,
            trg_schema = s.trg_schema,
            is_active = 1,
            priority = 1,
            load_mode = 'RECOMPUTE',
            align_path = NULL,
            dedup_order = NULL{nb_set}
        WHEN NOT MATCHED AND s.action = 'UPSERT' THEN INSERT (
            id, pl_name, src_schema, src_tbl, trg_schema, trg_tbl,
            is_active, priority, load_mode, align_path, dedup_order{nb_col}
        ) VALUES (
            COALESCE((SELECT MAX(id) FROM {cfg}), 0) + s.seq,
            'PL_VV_TRANS_SLV_TO_SLV_1H', s.src_schema, s.src_tbl, s.trg_schema, s.trg_tbl,
            1, 1, 'RECOMPUTE', CAST(NULL AS STRING), CAST(NULL AS STRING){nb_val}
        )
    """)
    return int(missing), int(retired)


PL = "PL_VV_TRANS_SLV_TO_SLV_1H"
SETUP_EXEC = str(uuid.uuid4())
ensure_lock_row(PL, current_notebook_name("NB_SETUP_POI_AMENITY"))
ok, who = acquire_flow_lock(PL, SETUP_EXEC, SETUP_EXEC, LOCK_TIMEOUT_DEFAULT_MIN)
if not ok:
    raise ConfigError(f"Luồng {PL} đang chạy ({who}); chạy setup sau")
status = "FAILED"
try:
    EDGES_INSERTED, EDGES_RETIRED = with_retry(_write_edges, "edges slv_poi_amenity")
    status = "SUCCESS"
    print(f"edges_inserted={EDGES_INSERTED} edges_retired={EDGES_RETIRED}")
finally:
    release_flow_lock(PL, SETUP_EXEC, status)
```
